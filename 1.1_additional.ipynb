### Cut CSV in half

```python
import os
from pathlib import Path
import polars as pl
```

```python
dataset_name = "2022_place_canvas_history"
dataset_dir = Path("D:/Downloads/task_1_dataset")
formats_dir = Path("D:/Downloads/task_1_dataset/formats")
os.makedirs(formats_dir, exist_ok=True)
```

```python
dataset = pl.scan_csv(dataset_dir / f"{dataset_name}.csv", has_header=True, low_memory=True)

n_rows = dataset.select(pl.len()).collect().item()
first_half = dataset.slice(0, n_rows // 2)

first_half.sink_csv(dataset_dir / f"{dataset_name}_half.csv")
```
